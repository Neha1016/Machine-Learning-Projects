# Import Library

```python
import pandas as pd
import joblib
```

# lode models

```python
recommend_model = joblib.load("../models/crop_recommendation_model.pkl")

yield_model = joblib.load("../models/yield_prediction_model.pkl")
```

lode all encoders

```python
label_encoder = joblib.load("../models/label_encoder.pkl")

state_encoder = joblib.load("../models/state_encoder.pkl")

district_encoder = joblib.load("../models/district_encoder.pkl")

season_encoder = joblib.load("../models/season_encoder.pkl")

crop_encoder = joblib.load("../models/crop_encoder.pkl")
```

Load Profit Dataset

```python
url = "https://raw.githubusercontent.com/shreyzo/Crop-yield-and-profitability-prediction/main/datafile.csv"

df_profit = pd.read_csv(url)
```

```python
df_profit.head()
```

From modual 2 (yield_prediction)

```python
df_profit = df_profit.rename(columns={
    "Cost of Cultivation (`/Hectare) A2+FL": "Cost_Cultivation_A2FL",
    "Cost of Cultivation (`/Hectare) C2": "Cost_Cultivation_C2",
    "Cost of Production (`/Quintal) C2": "Cost_Production_C2",
    "Yield (Quintal/ Hectare) ": "Yield_Quintal_Hectare",
    "Support price": "Support_Price"
})
```

```python
# Revenue per Hectare

df_profit["Revenue_per_Hectare"] = (
    df_profit["Support_Price"] *
    df_profit["Yield_Quintal_Hectare"]
)
```

```python
# Profit per Hectare

df_profit["Profit_per_Hectare"] = (
    df_profit["Revenue_per_Hectare"] -
    df_profit["Cost_Cultivation_C2"]
)
```

recommend_crop function from Module 1 (crop_recommendation)

```python
def recommend_crop(N, P, K, temperature, humidity, ph, rainfall):
    

    input_data = pd.DataFrame({
        "N": [N],
        "P": [P],
        "K": [K],
        "temperature": [temperature],
        "humidity": [humidity],
        "ph": [ph],
        "rainfall": [rainfall]
    })

    
    prediction = recommend_model.predict(input_data)

    
    crop = label_encoder.inverse_transform(prediction)

    return crop[0]
```

from modual 2 

```python
def predict_yield(
    state,
    district,
    year,
    season,
    crop,
    temperature,
    humidity,
    soil_moisture
):

    try:
        state = state_encoder.transform([state])[0]
        district = district_encoder.transform([district])[0]
        season = season_encoder.transform([season])[0]
        crop = crop_encoder.transform([crop])[0]

    except ValueError:
        return None

    input_data = pd.DataFrame({
        "State_Name_encoded": [state],
        "District_Name_encoded": [district],
        "Season_encoded": [season],
        "Crop_encoded": [crop],
        "Crop_Year": [year],
        "Temperature": [temperature],
        "Humidity": [humidity],
        "Soil_Moisture": [soil_moisture]
    })

    prediction = yield_model.predict(input_data)

    return round(float(prediction[0]), 2)
```

from modual 3 predict_yield() function 

```python
def get_profit_estimate(crop, state):
    # Convert crop name to uppercase
    crop = crop.upper()

    # Filter matching row
    result = df_profit[
        (df_profit["Crop"] == crop) &
        (df_profit["State"] == state)
    ]

    # If no matching row
    if result.empty:
        return None

    # Get first matching row
    row = result.iloc[0]

    # Return required information
    return {
        "Crop": row["Crop"],
        "State": row["State"],
        "Support Price": row["Support_Price"],
        "Cost of Cultivation (C2)": row["Cost_Cultivation_C2"],
        "Historical Yield": row["Yield_Quintal_Hectare"],
        "Profit per Hectare": row["Profit_per_Hectare"]
    }
```

# Smart Advisory Function

```python
def smart_crop_advisory(
    N,
    P,
    K,
    temperature,
    humidity,
    ph,
    rainfall,
    state,
    district,
    year,
    season,
    soil_moisture
):

    print("=" * 60)
    print("SMART CROP ADVISORY SYSTEM")
    print("=" * 60)

    # ==========================
    # Step A : Recommend Crop
    # ==========================

    recommended_crop = recommend_crop(
        N,
        P,
        K,
        temperature,
        humidity,
        ph,
        rainfall
    )

    print(f"Recommended Crop : {recommended_crop}")

    # ==========================
    # Step B : Convert for Module 2
    # ==========================

    crop_for_yield = recommended_crop.strip().title()

    # ==========================
    # Step C : Predict Yield
    # ==========================

    predicted_yield = predict_yield(
        state=state,
        district=district,
        year=year,
        season=season,
        crop=crop_for_yield,
        temperature=temperature,
        humidity=humidity,
        soil_moisture=soil_moisture
    )

    if predicted_yield is None:
        print("Predicted Yield : Not Available")
    else:
        print(f"Predicted Yield : {predicted_yield:.2f}")

    # ==========================
    # Step D : Profit Estimation
    # ==========================

    profit_crop = profit_crop_mapping.get(recommended_crop)

    if profit_crop is None:

        profit = None

    else:

        profit = get_profit_estimate(
            profit_crop.lower(),
            state
        )

    # ==========================
    # Step E : Print Profit
    # ==========================

    if profit is None:

        print("Profit Information : Not Available")

    else:

        print(f"Support Price : ₹{profit['Support Price']}")
        print(f"Historical Yield : {profit['Historical Yield']}")
        print(f"Estimated Profit per Hectare : ₹{profit['Profit per Hectare']:.2f}")

    print("=" * 60)
```

# Integration Challenges and Solutions


Crop Name Conversion

Module 1 returns crop names in lowercase, so they were converted to Title Case before sending them to the Yield Prediction module.
Some crop names in the Profit Analysis dataset use different names, so a crop mapping dictionary was created to convert the recommended crop into the corresponding profit dataset crop name.

Handling Missing Crop Information

Some crops recommended by the Crop Recommendation model are not available in the Yield Prediction or Profit Analysis datasets.
The integration function checks for missing records before making predictions.
If information is unavailable, the system displays an appropriate message instead of producing an incorrect prediction or stopping execution.

Robust Error Handling

The integration function safely handles missing categories and unavailable data using conditional checks.
This ensures that the complete advisory system continues running even if one module cannot return a result.

Final Integration

The Smart Crop Advisory function combines Crop Recommendation, Yield Prediction, and Profit Analysis into a single workflow.
The function automatically performs all required data conversions before calling each module.
