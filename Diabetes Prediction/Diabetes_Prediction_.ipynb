## Part 1: Import Required Libraries

```python
# ==========================================
# Part 1 : Import Required Libraries
# ==========================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings("ignore")

%matplotlib inline
```

## Part 2: Load the Dataset

```python
# ==========================================
# Part 2 : Load Dataset
# ==========================================

df = pd.read_csv("diabetes.csv1.csv")

print("Dataset Shape :", df.shape)
df.head()
```

## Part 3: Understand the Dataset

```python
# Display dataset information
df.info()
```

```python
# Display column names
print("Columns:")
print(df.columns.tolist())

print("\nNumber of Rows    :", df.shape[0])
print("Number of Columns :", df.shape[1])
```

```python
# Statistical summary
df.describe()
```

## Part 4: Exploratory Data Analysis

```python
# Check missing values
missing_values = df.isnull().sum()
print(missing_values)
```

```python
# Check invalid zero values in medical measurements
zero_columns = ["Glucose", "BloodPressure", "SkinThickness", "Insulin", "BMI"]

zero_values = (df[zero_columns] == 0).sum()
print("Zero values in selected columns:")
print(zero_values)
```

```python
# Outcome distribution
print("Outcome Distribution:")
print(df["Outcome"].value_counts())
```

```python
# Visualize outcome distribution
plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="Outcome")
plt.title("Diabetes Outcome Distribution")
plt.xlabel("Outcome (0 = No Diabetes, 1 = Diabetes)")
plt.ylabel("Number of Patients")
plt.show()
```

```python
# Correlation heatmap
plt.figure(figsize=(10, 7))
sns.heatmap(df.corr(), annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

## Part 5: Handle Invalid/Missing Values

In this dataset, some medical measurements contain `0`, which is not a meaningful value for variables such as Glucose, BloodPressure, SkinThickness, Insulin and BMI.

These zero values are treated as missing values and replaced using simple statistical values.

```python
# Create a copy for data cleaning
clean_df = df.copy()

# Replace invalid zero values with NaN
clean_df[zero_columns] = clean_df[zero_columns].replace(0, np.nan)

# Fill missing values
clean_df["Glucose"] = clean_df["Glucose"].fillna(clean_df["Glucose"].mean())
clean_df["BloodPressure"] = clean_df["BloodPressure"].fillna(clean_df["BloodPressure"].mean())
clean_df["SkinThickness"] = clean_df["SkinThickness"].fillna(clean_df["SkinThickness"].median())
clean_df["Insulin"] = clean_df["Insulin"].fillna(clean_df["Insulin"].median())
clean_df["BMI"] = clean_df["BMI"].fillna(clean_df["BMI"].median())

print("Total Missing Values After Cleaning :", clean_df.isnull().sum().sum())
```

```python
# View cleaned dataset
clean_df.head()
```

## Part 6: Prepare Features and Target

```python
# Independent variables (features)
X = clean_df.drop("Outcome", axis=1)

# Dependent variable (target)
y = clean_df["Outcome"]

print("Feature Shape :", X.shape)
print("Target Shape  :", y.shape)
```

## Part 7: Train-Test Split

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Data Shape :", X_train.shape)
print("Testing Data Shape  :", X_test.shape)
```

## Part 8: Feature Scaling

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Feature scaling completed.")
```

## Part 9: Train Logistic Regression Model

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(random_state=42)

model.fit(X_train_scaled, y_train)

print("Logistic Regression model trained successfully.")
```

## Part 10: Model Prediction and Evaluation

```python
# Make predictions
y_pred = model.predict(X_test_scaled)

print("Predictions:")
print(y_pred)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print("Logistic Regression Accuracy : {:.2f}%".format(accuracy * 100))
```

```python
# Classification report
print("Classification Report:")
print(classification_report(y_test, y_pred))
```

```python
# Confusion matrix
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(6, 5))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["No Diabetes", "Diabetes"],
    yticklabels=["No Diabetes", "Diabetes"]
)

plt.title("Confusion Matrix - Logistic Regression")
plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")
plt.show()
```
